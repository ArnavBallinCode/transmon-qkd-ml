```python
import numpy as np

N = 20

alice_bits = np.random.randint(0, 2, N)

alice_bases = np.random.randint(0, 2, N)
bob_bases = np.random.randint(0, 2, N)

print("Alice Bits :", alice_bits)
print("Alice Bases:", alice_bases)
print("Bob Bases  :", bob_bases)
```

```python
state_labels = []

for bit, basis in zip(alice_bits, alice_bases):

    if basis == 0:  # Z basis

        if bit == 0:
            state_labels.append("|0>")
        else:
            state_labels.append("|1>")

    else:  # X basis

        if bit == 0:
            state_labels.append("|+>")
        else:
            state_labels.append("|->")

print(state_labels)
```

```python
bob_results = []

for bit, a_basis, b_basis in zip(
    alice_bits,
    alice_bases,
    bob_bases
):

    if a_basis == b_basis:

        bob_results.append(bit)

    else:

        bob_results.append(
            np.random.randint(0, 2)
        )

print(bob_results)
```

```python
sifted_alice = []
sifted_bob = []

for i in range(N):

    if alice_bases[i] == bob_bases[i]:

        sifted_alice.append(
            alice_bits[i]
        )

        sifted_bob.append(
            bob_results[i]
        )

print("Alice Key:", sifted_alice)
print("Bob Key  :", sifted_bob)
```

```python
errors = 0

for a, b in zip(
    sifted_alice,
    sifted_bob
):

    if a != b:
        errors += 1

qber = errors / len(sifted_alice)

print("QBER =", qber)
```
