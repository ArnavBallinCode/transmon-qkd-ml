```python
import numpy as np

N = 1000

alice_bits = np.random.randint(0, 2, N)

alice_bases = np.random.randint(0, 2, N)

bob_bases = np.random.randint(0, 2, N)

p_relax = 0.05
```

```python
noisy_bits = []

for bit in alice_bits:

    if bit == 1:

        if np.random.random() < p_relax:

            noisy_bits.append(0)

        else:

            noisy_bits.append(1)

    else:

        noisy_bits.append(0)

noisy_bits = np.array(noisy_bits)
```

```python
bob_results = []

for noisy_bit, a_basis, b_basis in zip(
    noisy_bits,
    alice_bases,
    bob_bases
):

    if a_basis == b_basis:

        bob_results.append(noisy_bit)

    else:

        bob_results.append(
            np.random.randint(0, 2)
        )
```

```python
sifted_alice = []
sifted_bob = []

for i in range(N):

    if alice_bases[i] == bob_bases[i]:

        sifted_alice.append(
            int(alice_bits[i])
        )

        sifted_bob.append(
            int(bob_results[i])
        )
```

```python
errors = 0

for a, b in zip(
    sifted_alice,
    sifted_bob
):

    if a != b:
        errors += 1

qber = errors / len(sifted_alice)

print("Errors:", errors)
print("Sifted Length:", len(sifted_alice))
print("QBER:", qber)
```
